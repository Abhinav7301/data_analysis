Nominal data consists of categories that do not follow any order or ranking.
These values act only as names or labels and cannot be sorted meaningfully. In the given dataset, variables like Gender, Loan_ID, Education, and Loan_Status are nominal because their values represent different categories without hierarchy.

Ordinal data represents categories with a meaningful order or ranking.
The difference between the levels is not measurable, but the order matters. In this dataset, the column Dependents is ordinal because the values (0, 1, 2, 3+) show increasing levels. Similarly, Credit_History can be considered ordinal since it indicates a ranked condition (bad → good).

Discrete data consists of numerical values that can be counted but not measured continuously.
They represent whole-number quantities without decimal values. In the dataset, columns such as ApplicantIncome, CoapplicantIncome, and Dependents are discrete because they represent countable values.

Continuous data represents measurable numeric values that can take any value within a range, including fractional or decimal values.
In this dataset, LoanAmount is continuous because its value can vary smoothly and is based on measurement.
